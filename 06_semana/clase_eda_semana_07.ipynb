## Deteccion de **OUTLIERS**


```python
df_numericos = df[['HP', 'Cylinders','Price']]
```

```python
# A donde se aplica?  columnas categoricas? o columnas numericas?
Q1 = df_numericos.quantile(q=0.25)
Q3 = df_numericos.quantile(q=0.75)
```

```python
IQR = Q3 - Q1
```

```python
df_numericos
```

```python
df_numericos[ ~(df_numericos < (Q1 - 1.5 * IQR) ) | (df_numericos > (Q3 + 1.5 * IQR))]
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df_numericos['Price'], orient="h")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df_numericos, x='Price', orient="h")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df_numericos['HP'], orient="h")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df_numericos['Cylinders'], orient="h")
plt.show()
```

```python
df = df[~((df_numericos < (Q1 - 1.5 * IQR)) | (df_numericos > (Q3 + 1.5 * IQR))).any(axis=1)]
print(f"\nDimensión del DataFrame tras remover outliers: {df.shape}")
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='Price', orient="h")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df['Cylinders'], orient="h")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df['Price'], orient="h")
plt.show()
```
