Databases -> Clase 1

```python
import sqlite3
import pandas as pd
```

```python
DB_PATH:str = "blog.sqlite"
```

```python
conx = sqlite3.connect(DB_PATH)
```

```python
user_table_ddl:str = '''
CREATE TABLE IF NOT EXISTS user (
    id INTEGER PRIMARY KEY,
    username TEXT NOT NULL,
    password TEXT NOT NULL
)
'''
```

```python
cursor_sqlite = conx.cursor()
```

```python
query = cursor_sqlite.execute("SELECT datetime('now') AS data;")
```

```python
query.fetchall()
```

```python
cursor_sqlite.execute(user_table_ddl)
conx.commit()
```

```python
user_insert_dml:str = '''
INSERT INTO user (id, username, password)
VALUES
(1, 'admin', 'admin'),
(2, 'pirulo', 'pirulino');
'''
```

```python
cursor_sqlite.execute(user_insert_dml)
conx.commit()
```

```python
cursor_sqlite.execute('SELECT * FROM user')
cursor_sqlite.fetchall()
```

```python
# Query a table and display it interactively
df = pd.read_sql_query("SELECT * FROM user", conx)
df
```
