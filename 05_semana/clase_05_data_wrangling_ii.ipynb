```python
query = "SELECT * FROM ecommerce_transacciones"
```

# deduplicacion de datos

```python
df_ecommerce.duplicated()
df_ecommerce.drop_duplicates(inplace=True)
df_ecommerce
```

# Transformacions de datos

```python
DOLAR_OFICIAL:int = 1_515
PERCEPCION:float = 0.3
IVA:float = 1.21
```

```python
def aplicacion_cambio_a_pesos(monto: float) -> float:
    return monto * DOLAR_OFICIAL - monto * PERCEPCION - monto * IVA
```

```python
df_ecommerce['monto_pesos_argentinos'] = df_ecommerce['monto_usd'].apply(aplicacion_cambio_a_pesos)
```

```python
df_ecommerce.head()
```

```python
df_ecommerce.estado_envio.value_counts()
```

```python
dict_datos: dict = {
    "Entregado": 1,
    "En tránsito": 0,
    "Devuelto": 2
}
```

```python
df_ecommerce['estado_envio'] = df_ecommerce['estado_envio'].map(dict_datos)
```

## Reporte de ventas

```python
reporte_ventas = pd.pivot_table(
    data=df_ecommerce,
    values='monto_usd',
    index='cliente_zona',
    columns='categoria',
    aggfunc='mean'
)
```

```python
reporte_ventas
```

```python
reporte_plano = reporte_ventas.reset_index()
```

```python
reporte_plano
```

```python
pd.melt(
    reporte_plano,
    id_vars = ['cliente_zona'],
    value_vars= "Electrónica,Hogar,Ropa".split(','),
    var_name='categoria',
    value_name='ticket_promedio'
)
```

### Dataframe -> melt y pivot

```python
df_ventas = pd.DataFrame({
    'Sucursal': ['A', 'B', 'C'],
    'Enero': [100, 150, 200],
    'Febrero': [120, 160, 220],
    'Marzo': [110, 170, 210]
})

df_ventas
```

```python
df_vertical = pd.melt(
    df_ventas, #  data que yo trabajar
    id_vars=['Sucursal'],  # col que se mantiene fija
    value_vars=['Enero', 'Febrero', 'Marzo'], # col que se convierten en filas
    var_name='Mes Apilado', # nombre de columna de variables
    value_name='Ventas' # nombre de la columna con los valores
)
```

```python
df_vertical
```

```python
df_vertical.pivot(
    index='Sucursal',
    columns='Mes Apilado',
    values='Ventas'
)
```
