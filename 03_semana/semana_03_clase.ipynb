Crear conexion con database

```python
engine = create_engine(
    DATABASE_URL.format(
        DATABASE_PWD=userdata.get('DATABASE_PWD')
    )
)
```

```python
query:str = '''
    SELECT
    c.customername, c.city, o.amount
    FROM customers AS c
    JOIN orders AS o  -- ON c.customerid = o.customerid
    USING(customerid)
    WHERE
        c.country  ILIKE 'mexico' -- intrinseco de PG
    AND o.amount > 200;
'''
```

```python
with engine.connect() as conn:
    df = pd.read_sql_query(sql=text(query), con=conn)
```

RestCountries

```python
url:str =  "https://api.restcountries.com/countries/v5?region=Europe"
auth:str =  f"Bearer {userdata.get('REST_API_KEY')}"
```

```python
response = requests.get(url, headers={'Authorization':auth})
```

```python
data = response.json()
```

```python
paises_data = data['data']['objects']
```

```python
def clean_response(pais:dict) -> dict:
    nombre = pais.get('names', {}).get('common','Desconocido')
    poblacion = pais.get('population', 0)
    capitales = pais.get('capitals', [])
    capital = ", ".join([c.get('name', '') for c in capitales]) if capitales else "Sin capital"
    monedas = pais.get('currencies', [])
    moneda = ", ".join([m.get('name', '') for m in monedas]) if monedas else "Sin moneda"

    return {
        'nombre': nombre,
        'poblacion': poblacion,
        'capital': capital,
        'moneda': moneda
    }
```

```python
df_paises  = pd.DataFrame([clean_response(pais) for pais in paises_data])
```
